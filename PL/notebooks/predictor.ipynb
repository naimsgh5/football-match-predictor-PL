# M1 — Raw data validation (Premier League)

Checks before any feature engineering: row count, missing values, team name consistency, per-season completeness.

```python
import pandas as pd

df = pd.read_csv("../data/raw/premier_league_results.csv", parse_dates=["date"])
df = df.sort_values("date").reset_index(drop=True)

print(f"{len(df)} matches, from {df['date'].min().date()} to {df['date'].max().date()}")
df.head()
```

```python
# Missing values
print(df.isna().sum())
```

```python
# Unique teams on the home side vs away side -> should be identical
home_teams = set(df["home_team"].unique())
away_teams = set(df["away_team"].unique())
print(f"{len(home_teams)} home teams, {len(away_teams)} away teams")
print("Diff:", home_teams.symmetric_difference(away_teams))
```

```python
# Number of matches per season (Aug -> May) -> should be close to 380 (20 teams) for a complete season
def get_season(d):
    return d.year if d.month >= 8 else d.year - 1

df["season"] = df["date"].apply(get_season)
df.groupby("season").size()
```

## M1 conclusion

- 1900 matches, 2021-08-13 → 2026-05-24, 0 missing values.
- 27 distinct teams (expected: more than 20 over 5 seasons due to promotion/relegation), names consistent between home and away (no discrepancy).
- **5 complete seasons (380 matches each)**, including 2025/26 after completion via `data/raw/E0_2025_26_footballdata.csv` (football-data.co.uk) — the 21 missing matches were added, 0 score discrepancies found on matches already present.

Data validated, ready for M2 (feature engineering).
