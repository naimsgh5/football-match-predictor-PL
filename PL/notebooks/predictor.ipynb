# M1 — Validation des données brutes (Premier League)

Contrôles avant tout feature engineering : volumétrie, valeurs manquantes, cohérence des noms d'équipes, complétude par saison.

```python
import pandas as pd

df = pd.read_csv("../data/raw/premier_league_results.csv", parse_dates=["date"])
df = df.sort_values("date").reset_index(drop=True)

print(f"{len(df)} matchs, du {df['date'].min().date()} au {df['date'].max().date()}")
df.head()
```

```python
# Valeurs manquantes
print(df.isna().sum())
```

```python
# Equipes uniques cote domicile vs exterieur -> doivent etre identiques
home_teams = set(df["home_team"].unique())
away_teams = set(df["away_team"].unique())
print(f"{len(home_teams)} equipes domicile, {len(away_teams)} equipes exterieur")
print("Diff:", home_teams.symmetric_difference(away_teams))
```

```python
# Nombre de matchs par saison (aout -> mai) -> doit etre proche de 380 (20 equipes) pour une saison complete
def get_season(d):
    return d.year if d.month >= 8 else d.year - 1

df["season"] = df["date"].apply(get_season)
df.groupby("season").size()
```

## Conclusion M1

- 1900 matchs, 2021-08-13 → 2026-05-24, 0 valeur manquante.
- 27 équipes distinctes (normal : plus de 20 sur 5 saisons à cause des promotions/relégations), noms cohérents entre domicile et extérieur (aucune divergence).
- **5 saisons complètes (380 matchs chacune)**, y compris 2025/26 après complétion via `data/raw/E0_2025_26_footballdata.csv` (football-data.co.uk) — les 21 matchs manquants ont été ajoutés, 0 divergence de score détectée sur les matchs déjà présents.

Données validées, prêtes pour M2 (feature engineering).
